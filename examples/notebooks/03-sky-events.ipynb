# Events in the sky

Eclipses, a solar return, rise and set, and gochara (transits counted from the natal Moon and lagna), all from the
built-in ephemeris.

```python
from eshana import ChartEngine, Context

eng = ChartEngine()
for e in eng.eclipses("2026-01-01T00:00:00Z", "2027-01-01T00:00:00Z")["eclipses"]:
    print(e["max_utc"][:16], e["kind"], e["type"])
```

```python
ctx = Context.from_local_birth("1990-05-15", "14:30", "Asia/Kolkata", 28.6139, 77.2090)
r = eng.returns(ctx, "2026-01-01T00:00:00Z", "2027-01-01T00:00:00Z", body="sun")
print("solar return:", r["returns"][0]["utc"])
```

```python
day = eng.rise_set(28.6139, 77.2090, "2026-11-24", "Asia/Kolkata")["days"][0]
for b in day["bodies"]:
    print(b["body"], b["status"], (b.get("rise") or {}).get("local", "-")[11:19], (b.get("set") or {}).get("local", "-")[11:19])
```

```python
g = eng.gochara(ctx, "2026-01-01T00:00:00Z", "2027-01-01T00:00:00Z", bodies=["saturn", "jupiter"])
for t in g["transits"]:
    for s in t["segments"]:
        print(t["body"], s["start_utc"][:10], "sign", s["sign"], "house from Moon", s["house_from_moon"])
```
