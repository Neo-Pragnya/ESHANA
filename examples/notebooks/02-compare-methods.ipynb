# Comparing methods

The same birth under two configurations: two ayanāṃśas, then two house systems. The engine reports every value that
changed and whether a sign changed, so a choice of method is visible instead of hidden.

```python
from eshana import ChartEngine, Context

ctx = Context.from_local_birth("1990-05-15", "14:30", "Asia/Kolkata", 28.6139, 77.2090)
eng = ChartEngine()
c = eng.compare(ctx, "vedic", {"ayanamsha": "lahiri"}, {"ayanamsha": "raman"})
print(c["summary"])
for ch in c["changes"][:5]:
    print(f"{ch['id']:<10} {ch.get('a_sign', ''):<10} -> {ch.get('b_sign', ''):<10} {ch.get('difference_deg', 0):+.3f}")
```

```python
w = eng.compare(ctx, "western", {"house_system": "placidus"}, {"house_system": "koch"}, zodiac="tropical")
print(w["summary"])
```

Every ayanāṃśa and house system the engine knows is listed by `eshana methods` (and `ChartEngine.info()`).
