# A first chart

One birth, one Vedic chart: positions, a derived attribute with the rule that made it, the provenance, and what
happens when an input is missing. `pip install eshana` is all it needs; nothing is downloaded.

```python
from eshana import ChartEngine, Context, EshanaError

ctx = Context.from_local_birth("1990-05-15", "14:30", "Asia/Kolkata", 28.6139, 77.2090)
eng = ChartEngine()
chart = eng.vedic_chart(ctx, ayanamsha="lahiri", divisions=[1, 9], dasha=True)
for p in chart.raw["positions"][:9]:
    print(f"{p['id']:<10} {p['sign']['name']:<10} {p['longitude_deg']:8.4f}")
```

Every derived value names its rule and what it was derived from:

```python
nak = next(a for a in chart.raw["derived_attributes"] if a["id"] == "vedic.moon.nakshatra")
print(nak["value"]["name"], nak["value"]["pada"], "|", nak["rule_version"], "| from", nak["derived_from"])
print(chart.provenance["assumptions"][:3])
```

The engine refuses rather than guesses. A local time that the clocks skipped is an error with a stable code:

```python
try:
    eng.vedic_chart(Context.from_local_birth("2024-03-10", "02:30", "America/New_York"), ayanamsha="lahiri")
except EshanaError as e:
    print(e.code)
```

In Jupyter, a `Chart` shows the full report inline; `chart.to_html()` returns it as text.

```python
html = chart.to_html(sections=["overview"])
print(len(html) > 1000)
```
